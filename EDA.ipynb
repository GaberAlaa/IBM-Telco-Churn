```python
removed_col = ['CustomerID','Count','Country','State','Lat Long','Latitude','Longitude','Zip Code','Churn Score']
df = df.drop(columns=removed_col)
```

```python
print(df['Total Charges'].isna().sum())
df['Total Charges']= df['Total Charges'].fillna(0)
print(df['Total Charges'].isna().sum())
```

```python
def save_plot(filename):
    path = os.path.join(DASHBOARD_Pics_PATH, filename)
    plt.tight_layout()
    plt.savefig(path, dpi=150, bbox_inches="tight")
    plt.close()
    
    print(f"Saved: {path}")
```

```python
objects_col = df.select_dtypes(include='str').drop(columns=['City','Churn Reason']).columns

count_fig , axes = plt.subplots(4,5, figsize=(20,10))
axes = axes.flatten()
for i in range(objects_col.size):
    col_name = objects_col[i]
    sns.countplot(df[col_name], ax=axes[i])
    axes[i].set_title(f'count of {col_name}')

count_fig.tight_layout()
count_fig.show()
save_plot("graph_1")
```

```python
num_col = df.select_dtypes(include='number').columns

hist_fig , axes = plt.subplots(2,3, figsize=(20,10))
axes = axes.flatten()
for i in range(num_col.size):
    col_name = num_col[i]
    sns.histplot(df[col_name], ax=axes[i])
    axes[i].set_title(f'distribuation of {col_name}')

hist_fig.tight_layout()
hist_fig.show()
save_plot('graph_2')
```

```python
box_fig , axes = plt.subplots(2,3, figsize=(20,10))
axes = axes.flatten()
for i in range(num_col.size):
    col_name = num_col[i]
    sns.boxplot(df[col_name], ax=axes[i])
    axes[i].set_title(f'distribuation of {col_name}')

box_fig.tight_layout()
box_fig.show()
save_plot('graph_3')
```

```python
sns.countplot(data=df , x='Churn Label')
save_plot('graph_4')
```

```python
plt.figure(figsize=(6, 4))
df.groupby("Contract")["Churn Value"].mean().sort_values().plot(kind="barh")
plt.title("Churn Rate by Contract Type")
plt.xlabel("Churn Rate")

save_plot('graph_5')
```

```python
plt.figure(figsize=(5, 4))
sns.boxplot(x="Churn Label", y="Monthly Charges", data=df)
plt.title("Monthly Charges by Churn")
save_plot('graph_6')
```

```python
plt.figure(figsize=(6, 4))
df.groupby("Payment Method")["Churn Value"].mean().sort_values().plot(kind="bar")
plt.title("Churn Rate by Contract Type")
plt.xlabel("Payment Method")
save_plot('graph_7')
```

```python
reasons = (df[df["Churn Label"] == "Yes"]["Churn Reason"].value_counts().head(12).sort_values(ascending=False))

sns.barplot(x=reasons.values,y=reasons.index)
save_plot('graph_8')
```
